## Imports

```python
from dotenv import load_dotenv
import os
from src.logger import logger
import time

# Constants
from src.utils import RESULTS_DIR, JUDGE_INFO, MODELS_TO_TEST

# Functions
from src.utils import get_safe_filename, run_grouped_test, run_grouped_judge

# Env
load_dotenv()
```

## Execution

```python
logger.info(f"PHASE 1: GENERATING RESPONSES FOR {len(MODELS_TO_TEST)} MODELS")

for provider, target_model in MODELS_TO_TEST:
    safe_name = get_safe_filename(provider, target_model)
    output_file = f"{RESULTS_DIR}/{safe_name}_grouped.jsonl"
    
    logger.info(f"\n[RUNNING] {target_model} -> {output_file}")
    print(f"\n[RUNNING] {target_model} -> {output_file}")
    run_grouped_test(provider, target_model, output_file, limit=None)
    
    # Small sleep to avoid aggressive rate limiting between models
    time.sleep(2)

logger.info("PHASE 1 COMPLETE.")
```

```python
logger.info(f"PHASE 2: JUDGING RESPONSES USING {JUDGE_INFO[1]}")

for provider, target_model in MODELS_TO_TEST:
    safe_name = get_safe_filename(provider, target_model)
    input_file = f"{RESULTS_DIR}/{safe_name}_grouped.jsonl"
    judged_file = f"{RESULTS_DIR}/{safe_name}_revised.jsonl"
    
    if os.path.exists(input_file):
        logger.info(f"\n[JUDGING] {target_model}...")
        run_grouped_judge(input_file, judged_file)
    else:
        logger.info(f"\n[SKIP] {input_file} not found.")

logger.info("PHASE 2 COMPLETE.")
```
